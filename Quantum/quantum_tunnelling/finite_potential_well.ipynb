# **Quantum Tunneling** - Finite Potential Well

In this project we discuss the classical quantum mechanics problem of the finite potential well. We begin with the TDSE (time dependent schrodinger equation), which is:
$$

  i\hbar \frac{\partial}{\partial t}\psi(\mathbf{r}, t) = \left[ -\frac{\hbar^2}{2m} \nabla^2 + V(\mathbf{r}, t) \right] \psi(\mathbf{r}, t)

$$
where $ \mathbf{r} \in \mathbb{R}^3, t > 0 $ and $ \psi: \mathbb{R}^3 \times \mathbb{R}_{\geq 0} \to \mathbb{C} $ is a so-called wavefunction representing the motion of a particle

We proceed by seperation of variables. Assume our wavefunction satisfies $ \psi(\mathbf{r}, t) = \phi(\mathbf{r}) \rho(t)$, ($ \phi:\mathbb{R}^3 \to \mathbb{C}, \rho:\mathbb{R}_{\geq 0} \to \mathbb{C}$) i.e we can seperate/decompose it into a function of only position, and a function of only time. Substituting in equation above gives:
$$
  i\hbar \phi (\mathbf{r}) \frac{\mathrm{d}\rho(t)}{\mathrm{d}t} = \rho(t) \left[ -\frac{\hbar^2}{2m} \nabla^2 \phi(\mathbf{r}) + V(\mathbf{r}, t) \phi(\mathbf{r}) \right] 
$$

Then, dividing through by $ \phi(\mathbf{r}) \rho(t) $ gives:
$$
\frac{i\hbar}{\rho(t)}  \frac{\mathrm{d}\rho(t)}{\mathrm{d}t} = \frac{1}{\phi({\mathbf{r}})} \left[ -\frac{\hbar^2}{2m} \nabla^2 \phi(\mathbf{r}) + V(\mathbf{r}, t) \phi(\mathbf{r}) \right] 
$$

In our problem, $ V(\mathbf{r}, t) = V(\mathbf{r})$, i.e no time dependece. This gives the LHS as a function of time, RHS a function of position, so both must be constant. We call this constant E. Hence, solving for $\rho(t)$ we have:
$$
\frac{i\hbar}{\rho(t)}  \frac{\mathrm{d}\rho(t)}{\mathrm{d}t} = E \implies \frac{1}{\rho(t)} \frac{\mathrm{d}\rho(t)}{\mathrm{d}t} = - \frac{i E}{\hbar}
$$
So we get solution $\rho(t) = \rho(0) e^{{- \frac{i E t}{\hbar}}}$. Solving for $\phi(\mathbf{r})$:
$$
-\frac{\hbar^2}{2m} \nabla^2 \phi(\mathbf{r}) + V(\mathbf{r}) \phi(\mathbf{r}) = E \phi(\mathbf{r}) \iff \hat{H} \phi = E \phi
$$
where $\hat{H}$ is the Hamiltonian operator and defined here by:
$$
\hat{H} = -\frac{\hbar^2}{2m} \nabla^2  + V(\mathbf{r}) 
$$
This is called the time-independent Schrodinger equation. We study the finite potential well, which we define as the region $X = [-a, a] \times [-b, b] \times [-c, c]$ for $a,b,c >0$. For $ V_0 > 0 $, the potential we are working with here is:

\begin{cases} 
0 & \text{if } r \in X, \\
V_0 & \text{if } r \notin X
\end{cases}

We begin by solving this in $\mathbb{R}$, before generalising to $\mathbb{R}^3$, including time dependence along the way.

## Static solution in $\mathbb{R}$:

```python
from solving import potential_well_1d, solve_tise_1d #importing 1d solvers from the accompanying files
import numpy as np
```

```python
#parameters we're working with
x_grid = np.linspace(-8, 8, 1000)
V_0 = 100.0
a = 2.0

#potential, eigenvalues, eigenvectors (often referred to as eigenfunctions/eigenstates)
V_1d = potential_well_1d(x_grid, V_0, a=a)
E_1d, phi_1d = solve_tise_1d(lambda x: potential_well_1d(x, V_0, a=a), x_grid, V_0)

print(f"Number of bound states: {len(E_1d)}")
```

```python
from plotting import plot_for_1d_stationary

plot_for_1d_stationary(V0=V_0, phi_1d=phi_1d, E_1d=E_1d, x_grid=x_grid, a=a)
```

This is the first contradiction from classical physics. The particle should not have any probabilty of crossing the boundary, since if it did, it would have kinetic energy $ T = E - V_0 < 0 $, clearly a contradiction. However in Quantum mechanics, the energy levels are quantised, as we see with the energy levels and no particles existing with energies in between our energy levels, and the corresponding wavefunctions only have to satisfy boundary conditions at $\pm a$ and satisfy TISE. The wavefunction decays outside the well at a rate of $ \kappa = \frac{\sqrt{2 m (V_0 - E)}}{\hbar} $, which becomes very small for the energy levels closer to the boundary, meaning those wavefunctions decay slower outside the well, as we can see. Such wavefunctions also have a higher probability of 'jumping' the well too, which is proportional to $ e^\frac{-4a \sqrt{2m (V_0 - E)}}{\hbar} $, so as $ V_0 - E \to 0$, we get higher and higher probabilities, as evidenced above. These formulas can be shown by applying boundary conditions to the 1D TISE for this problem, alongside necessary continuity of the wavefunction and its derivatives at the boundary. We show this relationship between the probability of the particle being outside the well compared to its energy below.

```python
from plotting import plot_probability_outside_vs_energy

plot_probability_outside_vs_energy(phi_1d=phi_1d, E_1d=E_1d, x_grid=x_grid, a=a, V0=V_0)
```

We see the probability increases with energy, as expected. The reason the function is simply connected from point to point is because of the discretisation of energy levels.

The tails outside the well theoretically decay proportionally to $e^{ - \kappa d}$ (where the formula above for probability of existing originates from), where the symbol $\kappa =  \frac{\sqrt{2m (V_0 - E)}} {\hbar}$, and $d$ is the distance of the wave from the boundary. We show this on a plot below, where we plot the natural logarithm of both.

```python
from plotting import plot_exponential_tail_comparison

state = 12 #choose any bound state (state less than highest n above)

plot_exponential_tail_comparison(phi_n=phi_1d[:, state], E_n=E_1d[state], x_grid=x_grid, a=a, V0=V_0)
```

We see they match identically, matching the theory to reality.

## Time-dependent solution in $\mathbb{R}$

We now consider time dependence in 1D, before considering the 3D case.

```python
from plotting import plot_1d_time_evolution

state = 14
phi_n = phi_1d[:, state]
E_n = E_1d[state]

fig_time = plot_1d_time_evolution(phi_n=phi_n, E_n=E_n, x_grid=x_grid, t_max=10.0, N_t=300)
```

We see a sinusoidal wave pattern as the wave progresses in time, with period $E_n$ as expected by the complex exponential term gained from time dependence (and scaling to make $\hbar=1$). Substituting different states, for lower energies we see the plot stay pretty firmly between $[-2, 2]$ in the x-direction, but for higher energy levels this branches out past it, as expected by the probabilities calculated above. We also see more oscilations for higher energy states, as expected, and notice if we had plotted $| \psi | ^2$ (probability density) it would be independent of time as the modulus of the complex exponetial (only time dependence in the wavefunction) is constant (=1). To see time dependence of the probability density, we need to look at a super position of states, that is:

$$
\psi = \frac{1}{\sqrt{2}} ( \phi_i e^\frac{-i E_i t}{\hbar} + \phi_j e^\frac{-i E_j t}{\hbar})
$$
for $i,j \in \mathbb{Z}$ valid bound states. Taking modulus squared of this superposition gives:

$$
| \psi | ^2 = \frac{1}{2}  \phi_i  ^2 + \frac{1}{2} \phi_j ^2 + \phi_i \phi_j \cos{(\frac{(E_j - E_i) t}{\hbar})}
$$
where we see the time dependence in the last term. We visualise this below.

```python
from plotting import plot_superposition_prob
```

```python
state1, state2 = 0, 18 #between 0, N-1, where N is top energy state as defined in static case
phi1, phi2 = phi_1d[:, state1], phi_1d[:, state2]
E1, E2 = E_1d[state1], E_1d[state2]

plot_superposition_prob(phi_1=phi1, E_1=E1, phi_2=phi2, E_2=E2, x_grid=x_grid, t_max=10, N_t=300)
```

Here we see oscillations in time, physically meaning that the probability density of a superposition of eigenstates oscillates in time. This is interpreted as the probability of this particle being found inside well decreasing in time, and then increasing again at a later point, so at certain points in time, this particle is more likely to quantum tunnel compared to other points in time.

## Static solution in $\mathbb{R}^3$

For the numerical 3D model, we use the separable potential $V(x,y,z) = V_x(x) + V_y(y) + V_z(z) $ (where each is the same potential in each direction) and seek a product solution $\phi(x,y,z)=\phi_x(x)\phi_y(y)\phi_z(z)$. The TISE then separates into three 1D equations:
$$ 
\left[ -\frac{\hbar^2}{2m} \frac{d^2}{dx^2} + V_x(x)\right]\phi_x=E_x \phi_x,
$$
and equivalent equations in $y$ and $z$. The 3D eigenstates and energies are therefore
$$
\phi_{n_x,n_y,n_z}(x,y,z)=\phi_{n_x}(x)\phi_{n_y}(y)\phi_{n_z}(z),\qquad E_{n_x,n_y,n_z}=E_{n_x}^{(x)}+E_{n_y}^{(y)}+E_{n_z}^{(z)}.$$
This gives three Cartesian quantum numbers, $n_x,n_y,n_z$.

```python
from solving import solve_3d_well, compute_3d_state_seperable

X3, Y3, Z3, energies_3d, interps_3d = solve_3d_well(a=4.0, b=3.0, c=5.0, V0=10.0, N_1d=100, N_3d=100) #first 3 inputs are the grid for plot later(already wrapped inside that function)
E_x, E_y, E_z = energies_3d
phi_x_int, phi_y_int, phi_z_int = interps_3d
```

The individual 3D solutions are labelled by 3 quantum numbers $(n_x, n_y, n_z)$

```python
n_x, n_y, n_z = 1, 2, 2

E_3d = E_x[n_x] + E_y[n_y] + E_z[n_z]

phi_3d, prob_3d = compute_3d_state_seperable(X3, Y3, Z3, (phi_x_int, phi_y_int, phi_z_int), (n_x, n_y, n_z))
```

As previously mentioned, a particle is defined by one continuous wavefunction spread throughout $\mathbb{R}^3$. The wavefunction here has 3 standing wave patterns in the 3 co-ordinate directions $x, y, z$. When we pass through zero, we create nodal planes, which divide the continuous state into different regions of high probability. The green regions signify a region of high probability, light blue of low probability, and the dotted line the box/well we are working with. It can also be thought of having 3 quantum numbers, $n_x, n_y, n_z$ and varying them gives different possibilities of bound states. The number of distinct regions viewed in each direction is given by $n_i + 1$     $(i \in \set{x, y, z})$, making the total number of distinct regions $(n_x + 1)(n_y + 1)(n_z + 1)$. 

As we increase each quantum number $n_x, n_y, n_z$, the energy increases and so does the probability of the particle escaping the box, a pattern we have already seen in the 1D case.
