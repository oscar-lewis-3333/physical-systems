```python
#Defining the functions x_plus and x_minus using the solution given by the quadratic formula
from condition_numbers_a1 import x_plus, x_minus

#Defining the relative change functions for both x_plus and x_minus using the functions defined above.

from condition_numbers_a1 import R_plus, R_minus

#Printing appropiate results as requested in question
print("For p = -1, q = 1 and dp = ", 10**-3, " we have that the relative change of x_plus is", R_plus(-1, 10**-3, 1))
print("For p = 1, q = 1 and dp = ", 10**-3, " we have that the relative change of x_plus is", R_plus(1, 10**-3, 1))
print("For p = -1, q = 1 and dp = ", 10**-3, " we have that the relative change of x_minus is", R_minus(-1, 10**-3, 1))
print("For p = 1, q = 1 and dp = ", 10**-3, " we have that the relative change of x_minus is", R_minus(1, 10**-3, 1))
```

For $q = 1$, and $dp = 0.001$ the relative change for $x_{+}$ and $x_{-}$ is approximately the same, regardless of whether p $= \pm$ 1. This means we only consider $p \geq 0$ since $R(p,dp) \approx R(-p, dp)$

```python
#We define the values needed for part b.
values = np.linspace(-1, 1, 101)
y1_values = R_plus(values, 10**-3, 1)
y2_values = R_minus(values, 10**-3, 1)

#Defining K(p) for q = 1 since it is consistent for x_plus and x_minus

from condition_numbers_a1 import K
#defining the values of K(p)
k_values = K(values)

#Plotting R_plus and associated
plt.plot(values, y1_values, label='R_plus')
plt.plot(values, k_values, label='K(p)', linestyle='--')
plt.xlabel('p')
plt.legend()
plt.grid(True)
plt.show()

plt.plot(values, y2_values, label='R_minus')
plt.plot(values, k_values, label='K(p)', linestyle='--')
plt.xlabel('p')
plt.legend()
plt.grid(True)
plt.show()
```

The plots are approximately the same, and the corresponding $K(p)$ overlaps the $R_\pm$ plot entirely

```python
#Doing the same for q
from condition_numbers_a1 import R_plus2, R_minus2

print(R_plus2(1, 10**-3, 1))
print(R_plus2(1, 10**-3, -1))
print(R_minus2(1, 10**-3, 1))
print(R_minus2(1, 10**-3, -1))

#We now consider the case for q

values = np.linspace(-1, 1, 101)
y1_values = R_plus2(1, 10**-3, values)
y2_values = R_minus2(1, 10**-3, values)

#Defining K(q) for p = 1 since it is consistent for x_plus and x_minus
from condition_numbers_a1 import K_plus, K_minus

#defining the values of K(q)
k1_values = K_plus(values)
k2_values = K_minus(values)

#Plotting R_plus and associ
plt.plot(values, y1_values, label='R_plus')
plt.plot(values, k1_values, label='K(q)', linestyle='--')
plt.xlabel('q')
plt.legend()
plt.grid(True)
plt.show()

plt.plot(values, y2_values, label='R_minus')
plt.plot(values, k2_values, label='K(q)', linestyle='--')
plt.xlabel('q')
plt.legend()
plt.grid(True)
plt.show()
```

It is not trivial to see above, but K(q) does infact overlap with $R_{\pm}$ as shown in the plot for $K(q)$ apart from as $q \to -1$

```python


from lagrange_interpolation_a1 import lagrange_interp, q, m_value
grids = np.linspace(-5, 5, 101)

nodes_3, values_3 = m_value(3)
int_values_3 = lagrange_interp(grids, nodes_3, values_3)

nodes_5, values_5 = m_value(5)
int_values_5 = lagrange_interp(grids, nodes_5, values_5)

nodes_11, values_11 = m_value(11)
int_values_11 = lagrange_interp(grids, nodes_11, values_11)

nodes_21, values_21 = m_value(21)
int_values_21 = lagrange_interp(grids, nodes_21, values_21)

plt.plot(grids, int_values_3, label='Interpolation values for m = 3')
plt.plot(grids, q(grids), label='Real Function')
plt.xlabel('x')
plt.ylabel('function values')
plt.legend()
plt.grid(True)
plt.show()

plt.plot(grids, int_values_5, label='Interpolation values for m = 5')
plt.plot(grids, q(grids), label='Real Function')
plt.xlabel('x')
plt.ylabel('function values')
plt.legend()
plt.grid(True)
plt.show()

plt.plot(grids, int_values_11, label='Interpolation values for m = 11')
plt.plot(grids, q(grids), label='Real Function')
plt.xlabel('x')
plt.ylabel('function values')
plt.legend()
plt.grid(True)
plt.show()

plt.plot(grids, int_values_21, label='Interpolation values for m = 21')
plt.plot(grids, q(grids), label='Real Function')
plt.xlabel('x')
plt.ylabel('function values')
plt.legend()
plt.grid(True)
plt.show()
```

Here we observe that, as m increases, the approximation to the real function becomes more accurate, specifically as $|x|$ approaches 0. The approximation for $|x| \geq 4.6$ gets worse as m increases

```python
x_p = [1, 4.5]
print('The values of the lagrange interpolation for m = 3 at x = [1,4.5] are', lagrange_interp(x_p, nodes_3, values_3))
print('The values of the lagrange interpolation for m = 5 at x = [1,4.5] are',lagrange_interp(x_p, nodes_5, values_5))
print('The values of the lagrange interpolation for m = 11 at x = [1,4.5] are',lagrange_interp(x_p, nodes_11, values_11))
print('The values of the lagrange interpolation for m = 21 at x = [1,4.5] are',lagrange_interp(x_p, nodes_21, values_21))
print('The real function values at x = [1, 4.5] are', q(np.array(x_p)))
```

We see that for $x = 1$, the approximation improves as m increases, but for $x = 4.5$, the approximation gets worse until m = 21, and then is accurate entirely. In terms of Weierstrass's theorem, we see that at m = 21 we have found a polynomial which is within $\epsilon$ at $x = 1, 4.5$, hence showing, at least at these points, we have found a polynomial for which we have approximated a continuous function.
